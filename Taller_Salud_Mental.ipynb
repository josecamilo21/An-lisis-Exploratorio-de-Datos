```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
salud_mental = pd.read_csv("/content/drive/MyDrive/CIENCIA_DE_DATOS/Salud_Mental_20260406.csv")
```

```python
salud_mental.head(10)
```

```python
# PASO 1: Identifico las columnas de grupos de edad
# Excluyo las primeras 2 columnas (diagnóstico y código) y me quedo con el resto
columnas_edades = salud_mental.iloc[:, 2:].columns
```

```python
# PASO 2: Genero un duplicado del DataFrame para preservar los datos originales intactos
proporciones = salud_mental.copy() # Trabajaré sobre esta copia, dejando Salud_mental sin cambios
```

```python
# PASO 3: Convertir los valores de cada grupo de edad en proporciones
for col in columnas_edades:  # Itero sobre cada columna de grupo etario
    base = Salud_mental[col].sum()        # Acumulo el total de casos registrados en ese grupo
    proporciones[col] = Salud_mental[col] / base  # Normalizo cada valor dividiéndolo entre el total

# PASO 4: Inicializar la estructura donde almacenaré los hallazgos
hallazgos = {}  # Estructura clave-valor: grupo de edad → (enfermedad predominante, su proporción)

# PASO 5: Detectar qué enfermedad predomina en cada grupo etario
for col in columnas_edades:  # Itero nuevamente por cada grupo de edad
    pos_max = proporciones[col].idxmax()          # Localizo la fila con la proporción más elevada
    condicion = proporciones.loc[pos_max, 'diagnostico_ingreso']  # Recupero el diagnóstico de esa fila
    peso = proporciones.loc[pos_max, col]          # Extraigo el valor proporcional correspondiente
    hallazgos[col] = (condicion, peso)             # Registro el hallazgo en la estructura

# PASO 6: Presentar los hallazgos de forma legible
for edad, (condicion, peso) in hallazgos.items():  # Desempaco cada entrada del diccionario
    print(f"Grupo etario: {edad}")                          # Etiqueto el rango de edad
    print(f"Diagnóstico predominante: {condicion}")         # Muestro la enfermedad con mayor peso
    print(f"Proporción: {peso:.4f}")                        # Expreso la proporción con 4 decimales
    print("-" * 40)                                         # Separo visualmente cada bloque

# PASO 7 (OPCIONAL): Estructurar los hallazgos en una tabla
tabla_final = pd.DataFrame([
    {"Grupo etario": edad, "Diagnóstico": cond, "Proporción": val}
    for edad, (cond, val) in hallazgos.items()
])  # Convierto el diccionario en un DataFrame para una visualización más ordenada

tabla_final  # Despliego la tabla con los resultados consolidados
```
