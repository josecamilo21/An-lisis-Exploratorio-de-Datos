```python
import pandas as pd
```

```python
movie = pd.read_csv("https://raw.githubusercontent.com/Gustavo-Bruges/movies/refs/heads/main/movies.csv")
rating = pd.read_csv("https://raw.githubusercontent.com/Gustavo-Bruges/movies/refs/heads/main/ratings.csv")
```

```python
print(movie.shape)
print(rating.shape)
```

```python
print("-"*100)
print("movie dataset")
print(movie.head())

print("-"*100)
print("rating dataset")
print(rating.head())
```

```python
movie_ratings = movie.merge(rating, on= "id", suffixes=("_movie", "_rating"))
movie_ratings.head()
```

```python
movie_ratings.drop(["Unnamed: 0_movies", "Unnamed: 0_rating"], axis=1, inplace=True, errors='ignore')
```

```python
movie_ratings.head()
```

```python
print(movie_ratings.shape)
```

```python
movie_genres = pd.read_csv("https://raw.githubusercontent.com/Gustavo-Bruges/movies/refs/heads/main/movie_to_genres.csv")
```

```python
movie_genres.head()
```

```python
movie_genres.drop(["Unnamed: 0"], axis=1, inplace=True)
print(movie_genres.head())
```

```python
# 1. Buscamos el nombre real de la columna (por si es movie_id, ID, etc.)
col_izquierda = [c for c in movie.columns if 'id' in c.lower()][0]
col_derecha = [c for c in movie_genres.columns if 'id' in c.lower()][0]

# 2. Hacemos el merge usando esos nombres detectados
movie_generos = movie.merge(
    movie_genres,
    left_on=col_izquierda,
    right_on=col_derecha,
    suffixes=("_movie", "_genre")
)

# 3. Ver resultado
movie_generos.head()
```

```python
peliculas_crimen = movie_generos[movie_generos['genre'] == 'Crime']

crimen_famosa = peliculas_crimen.sort_values(by='popularity', ascending=False).iloc[0]

print("La película de crimen más famosa según tus datos es:")
print(f"Título: {crimen_famosa['title']}")
print(f"Popularidad: {crimen_famosa['popularity']}")
```

```python
sci_movie = movie_generos[movie_generos["genre"] == "Science Fiction"]
action_movie = movie_generos[movie_generos["genre"] == "Action"]
```

```python
print(sci_movie.head())
print(action_movie.head())
print(sci_movie.shape)
print(action_movie.shape)
```

```python
df_left = action_movie.merge(sci_movie, on="movie_id", how="left")
df_left.head()
```

```python
df_left = action_movie.merge(sci_movie, on="movie_id", how="left", suffixes=("_action", "_sci"))

print(df_left.shape)
```

```python
df_right = action_movie.merge(sci_movie, on="movie_id", how="right", suffixes=("_action", "_sci"))

print(df_right.shape)
```
